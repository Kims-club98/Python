```python
def creat_model():
    #학습데이터 불러오기
    import pandas as pd
    df_train=pd.read_csv('data/감성분석/train.csv')
    df_train=df_train[:1000]
    df_train.head()

    # 한글 형태소의 분리함수
    from konlpy.tag  import Okt
    okt=Okt()
    def okt_tokenizer(text):
        tokens=okt.morphs(text)
        return tokens
        
    #document 백터화
    from sklearn.feature_extraction.text import TfidfVectorizer
    vector=TfidfVectorizer(tokenizer=okt_tokenizer,ngram_range=(1,2),min_df=3,max_df=0.9,token_pattern=None)

    vector.fit(df_train['document'])
    vector_trans=vector.transform(df_train['document'])

    #모델생성 후 모델학습
    from sklearn.linear_model import LogisticRegression
    model=LogisticRegression(C=10,max_iter=100)
    model.fit(vector_trans, df_train['label'])
    return vector,model


```

```python
import re
vector,model=creat_model()
text='8기 인공지능(AI) 기반 피싱 사이트 제작, 신규 일반 최상의 메인활동임'
find_text=re.findall(r'[가-힣]+',text)
text_join=[' '.join(find_text)]
vector_text=vector.transform(text_join)
pred=model.predict(vector_text)
print(pred) #영화데이터 기반 1은 긍정/0은 부정

```
