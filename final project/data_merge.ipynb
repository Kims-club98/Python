```python
import pandas as pd
import os
```

```python

folder_path='data/공공데이터'
colums_to_use=['URL','기관명','기관 데이터포털에서 제공하는 목록명', '공공데이터포털에 등록된 목록명']

file_names=os.listdir(folder_path)
file_names=[f for f in file_names if f.endswith('.xlsx')]

df_list=[]

for file in file_names:
    try:
        file_path=os.path.join(folder_path, file)
        df=pd.read_excel(file_path,usecols=colums_to_use)
        df['출처파일']=file
        df_list.append(df)
    except Exception as e:
        print(f'{file}로딩실패: {e}')

df_all=pd.concat(df_list, ignore_index=True)
new_order=['기관명','기관 데이터포털에서 제공하는 목록명','공공데이터포털에 등록된 목록명','URL','출처파일']
df_all=df_all[new_order]
df_all.columns=['기관','기관제공목록명','정부24제공목록명','url','출처']
df_all.drop(columns=['출처'],inplace=True)

df_all.head(5)
df_all.to_csv('data/통합공공데이터.csv',sep=',')
```
