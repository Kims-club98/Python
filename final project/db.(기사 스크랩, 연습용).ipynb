```python
import requests
from bs4 import BeautifulSoup
import csv
import re
```

```python
save=open('data/scrap title.csv','w',encoding='utf-8',newline="")
write=csv.writer(save)
```

```python
#연합뉴스 기사제목 가지고오기

count=1
for page in range(1,21):
    url=f'https://www.yna.co.kr/news?page={page}&site=navi_latest_depth01'
    res=requests.get(url)
    print(res)
    soup=BeautifulSoup(res.text,'lxml')
    scraps=soup.find_all('span',attrs={'class':'title01'})
    upload=soup.find('span',attrs={'class':'txt-time'}).getText()
    for tag in scraps:
        # print(f"기사{index}: {tag.getText(strip=True)}")
        raw_tag=tag.getText(strip=True).replace('"','')
        clean_tag=re.sub(r'[\"\']','',raw_tag)
        count+=1
        data=[f"기사{count} 업로드:{upload}: {clean_tag}"]
        write.writerow(data)
```
